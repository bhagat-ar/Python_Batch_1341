Task1 -> you hve to read a number from console. Tell the user that the number is positive or negative.

```python
num= int(input("enter the number :"))

if num > 0:
    print(num,"number is positive")
elif num < 0 :
    print(num, "number is negative")
else:
    print(num,"number is zero")
```

Task 4: student exam grading system
write a program assigns a letter grade based on student's score .

```python
score= float(input("enter your exam score:"))

if score >= 90 and score <= 100:
    print("Grade : A")
elif score>= 80 and score < 90:
    print("Grade: B")
elif score >= 70 and score < 80:
    print("Grade : C")
elif score >= 60 and score < 70:
    print("Grade : D")
elif score >= 35 and score < 60:
    print("Grade : E")
else:
    print("failed")
```
