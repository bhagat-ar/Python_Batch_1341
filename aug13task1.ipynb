1) integer with all operators

```python
a= 10
b= 3
#Arithmetic operator
print("addition (a + b) :", a+b)
print("subtraction (a - b) :", a-b)
print("multiplication (a * b) :", a*b)
print("division (a/b) :", a/b)
print("floor division (a // b) :", a//b)
print("modulo (a % b) :", a% b)
print("exponentiation (a ** b) :", a**b)
print("**************************************")

#relational/comparison operator
print("Equal (==):", a==b)
print("not equal(!=) :", a!=b)
print("greater (>):", a > b)
print("less than (<):", a < b)
print("greater than equal (>=):", a>= b)
print("less than equal(<=):", a <= b)
print("**************************************")

#logical operator
print("a and b: ",a and b)
print("a or b: ",a or b)
print ("not a: ",not a)
print("**************************************")

#assignment operator
c=b
c+= 10
print ("c",c)
c+= 45 
print("c+= 45:",c )
c-= 5  
print("c-= 5 :", c)
c*=2
print("c*=2 : ", c)
c/=2
print("c/=2: ",c)
print("**************************************")

# bitwise operator
print("bitwise and (&) :", a & b)
print("bitwise or (|) :", a | b)
print("bitwise xor (^) :", a ^ b)
print("bitwise not (~) :",  ~ b)
print("bitwise left shift (<<) :", a << 2)
print("bitwise right shift (>>) :", a >> 2)


```
