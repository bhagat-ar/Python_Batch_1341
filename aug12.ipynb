2) Relational/ conditiona operator

```python
#1) relational operator(>,<, >=,<=,!=)

aman= 22
jay= 21
print(aman > jay)
print(aman != jay)
print(aman >= jay)
print(aman < jay)
print(aman <= jay)
```

```python
#2) conditional operator(if-else)

jay= 23
riya= 21
if jay >riya:
    print("jay will pay the bill")
else:
    print("riya will pay the bill")
```

5) bitwise operator(&, |, ^, !, <<,>>, >>>)

```python
#1) decimal 
a= 10
print (a, type(a))

#2) binary
v1= 101010
v2= 0B01020
print(v1, type(v1))
print (v1, type(v2))

#3) octal 
c1= 0o101010
c2= 0o678
print(c1, type(c1))
print (c1, type(c2))

#4) hexadecimal
var= 0xface
print(var, type(var))
```
