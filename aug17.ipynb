1) string slicing with step value/size:

```python
s= "Instagram"

print(s[2:6])
print(s[1:8])

print("*****************")
print(s[1:8:1])
print(s[1:8:2])
print(s[1:8:3])
print(s[8:1:-1])
print("*****************")

print(s[0:9:+1])
print(s[-1:-10:-1])
print(s[0::1])
print(s[0::2])
print("*****************")
print(s[::1])
print(s[::-1])
```

```python
email= "rahulsharma@gmail.com"
#1)
username= email[0:11:+1]
domain = email[13::+1]
print("username:", username.title())
print("domain:",domain)

print("*****************")
#2)
count = email.index("@")
print(count)

username= email[:count:+1]
domain = email[count+1::+1]
print("username:", username)
print("domain:",domain)
print("*****************")

#3)
username= email[:email.index("@"):+1]
domain = email[email.index("@")+1::+1]
print("username:", username)
print("domain:",domain)
print("*****************")
```

2) string inbuild methods/ functions

```python
file='myfile123.pdf'
count= file.index('.')
name= file[0:count]
extension= file[count+1 :]

print("name upper():", name.upper())
print("extension lower():", extension.lower())

print("name title() :", name.title())
print("name isalnum() :", name.isalnum())
print("name isascii():", name.isascii())
print("name is digit():", name.isdigit())
print("extension isspace():", extension.isspace())
print("extension isprintable():", extension.isprintable())
print("extension isidentifier():", extension.isidentifier())

print("**********************************************")

print("extension endswith():", file.endswith("pdf"))
print("extension startswith():", file.startswith("g"))

print("extension replace():", file.replace("myfile123", "arti23"))




```
